```python
import pandas as pd

url = "https://www.football-data.co.uk/mmz4281/2526/E0.csv"
df = pd.read_csv(url)

df = df[["Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR", "B365H", "B365D", "B365A"]]
print(df.head())
print(len(df), "matches")
```

```python
counts = df["FTR"].value_counts()
print(counts)

away_wins = counts["A"]
total = len(df)
print("Away win %:", away_wins / total * 100)
```

```python
home_scored = df.groupby("HomeTeam")["FTHG"].mean()
print(home_scored.sort_values(ascending=False))
```

```python
league_home_avg = df["FTHG"].mean()

home_attack = home_scored / league_home_avg
print(home_attack.sort_values(ascending=False).round(2))
```

```python
away_scored = df.groupby("AwayTeam")["FTAG"].mean()
print(away_scored.sort_values(ascending=False))
```

```python
league_away_avg = df["FTAG"].mean()

away_attack = away_scored / league_away_avg
print(away_attack.sort_values(ascending=False).round(2))
```

```python
home_conceded = df.groupby("HomeTeam")["FTAG"].mean()
print(home_conceded.sort_values(ascending=False))
```

```python
home_defence = home_conceded / league_away_avg
print(home_defence.sort_values().round(2))
```

```python
away_conceded = df.groupby("AwayTeam")["FTHG"].mean()
print(away_conceded.sort_values(ascending=False))
```

```python
away_defence = away_conceded / league_home_avg
print(away_defence.sort_values().round(2))
```

```python
strengths = pd.DataFrame({
    "home_attack": home_attack,
    "home_defence": home_defence,
    "away_attack": away_attack,
    "away_defence": away_defence,
}).round(2)

print(strengths)
```

```python
home_team = "Man City"
away_team = "Burnley"

home_xg = strengths.loc[home_team, "home_attack"] * strengths.loc[away_team, "away_defence"] * league_home_avg
away_xg = strengths.loc[away_team, "away_attack"] * strengths.loc[home_team, "home_defence"] * league_away_avg

print(home_team, "expected goals:", round(home_xg, 2))
print(away_team, "expected goals:", round(away_xg, 2))
```

```python
from scipy.stats import poisson

for goals in range(6):
    p = poisson.pmf(goals, home_xg)
    print(home_team, "score", goals, ":", round(p*100,1), "%")
```

```python
from scipy.stats import poisson

for goals in range(6):
    p = poisson.pmf(goals, away_xg)
    print(away_team, "score", goals, ":", round(p*100,1), "%")
```

```python
home_win = 0
draw = 0
away_win = 0

for h in range(10):
    for a in range(10):
        p = poisson.pmf(h, home_xg) * poisson.pmf(a, away_xg)
        if h > a:
            home_win += p
        elif h == a:
            draw += p
        else:
            away_win += p

print(home_team, "win:", round(home_win * 100, 1), "%")
print("Draw:", round(draw * 100, 1), "%")
print(away_team, "win:", round(away_win * 100, 1), "%")
```

```python
match = df[(df["HomeTeam"] == home_team) & (df["AwayTeam"] == away_team)]
print(match)
```

```python
odds_home = match["B365H"].iloc[0]
odds_draw = match["B365D"].iloc[0]
odds_away = match["B365A"].iloc[0]

print("Bookmaker", home_team, "win:", round(1 / odds_home * 100, 1), "%")
print("Bookmaker draw:", round(1 / odds_draw * 100, 1), "%")
print("Bookmaker", away_team, "win:", round(1 / odds_away * 100, 1), "%")
```

```python
total = 1 / odds_home + 1 / odds_draw + 1 / odds_away
print("Bookmaker total:", round(total * 100, 1), "%")

fair_home = (1 / odds_home) / total
fair_draw = (1 / odds_draw) / total
fair_away = (1 / odds_away) / total

print("Fair", home_team, "win:", round(fair_home * 100, 1), "%")
print("Fair draw:", round(fair_draw * 100, 1), "%")
print("Fair", away_team, "win:", round(fair_away * 100, 1), "%")
```

```python
def expected_goals(home, away):
    home_xg = strengths.loc[home, "home_attack"] * strengths.loc[away, "away_defence"] * league_home_avg
    away_xg = strengths.loc[away, "away_attack"] * strengths.loc[home, "home_defence"] * league_away_avg
    return home_xg, away_xg

print(expected_goals("Man City", "Burnley"))
print(expected_goals("Arsenal", "Chelsea"))
```

```python
h_xg, a_xg = expected_goals("Arsenal", "Chelsea")
print("Arsenal:", round(h_xg, 2), "| Chelsea:", round(a_xg, 2))
```

```python
def predict(home, away):
    home_xg, away_xg = expected_goals(home, away)

    home_win = 0
    draw = 0
    away_win = 0

    for h in range(10):
        for a in range(10):
            p = poisson.pmf(h, home_xg) * poisson.pmf(a, away_xg)
            if h > a:
                home_win += p
            elif h == a:
                draw += p
            else:
                away_win += p

    return home_win, draw, away_win
```

```python
home_win, draw, away_win = predict("Man United", "Chelsea")

print("Man United win:", round(home_win * 100, 1), "%")
print("Draw:", round(draw * 100, 1), "%")
print("Chelsea win:", round(away_win * 100, 1), "%")
```

```python
model_home = []
model_draw = []
model_away = []

for i in range(len(df)):
    home = df["HomeTeam"].iloc[i]
    away = df["AwayTeam"].iloc[i]

    hw, d, aw = predict(home, away)

    model_home.append(hw)
    model_draw.append(d)
    model_away.append(aw)

df["model_H"] = model_home
df["model_D"] = model_draw
df["model_A"] = model_away

print(df[["HomeTeam", "AwayTeam", "model_H", "model_D", "model_A"]].head(10).round(3))
```

```python
df["margin"] = 1 / df["B365H"] + 1 / df["B365D"] + 1 / df["B365A"]

df["book_H"] = (1 / df["B365H"]) / df["margin"]
df["book_D"] = (1 / df["B365D"]) / df["margin"]
df["book_A"] = (1 / df["B365A"]) / df["margin"]

print(df[["HomeTeam", "AwayTeam", "model_H", "book_H", "model_A", "book_A"]].head(10).round(3))
```

```python
df["edge_H"] = df["model_H"] - df["book_H"]

bets = df[df["edge_H"] > 0.05].copy()

bets["profit"] = -1.0
bets.loc[bets["FTR"] == "H", "profit"] = bets["B365H"] - 1

print("Number of bets:", len(bets))
print("Total profit (£1 per bet):", round(bets["profit"].sum(), 2))
print("ROI:", round(bets["profit"].sum() / len(bets) * 100, 1), "%")
```
